```
import pandas as pd

df = pd.read_csv('titanic.csv')

print(df.info())
print(df.head())

sex_mapping = {0: 'Male', 1: 'Female'}

df['sex'] = df['sex'].map(sex_mapping)

df['sex'] = df['sex'].astype('category')

embarked_mapping = {
    'C': 'Cherbourg',
    'Q': 'Queenstown',
    'S': 'Southampton'
}
df['embarked'] = df['embarked'].map(embarked_mapping)
df['embarked'] = df['embarked'].astype('category')

survived_mapping = {0: 'No', 1: 'Yes'}
df['survived'] = df['survived'].map(survived_mapping)
df['survived'] = df['survived'].astype('category')

pclass_mapping = {1: '1st', 2: '2nd', 3: '3rd'}
df['pclass'] = df['pclass'].map(pclass_mapping)

df['pclass'] = pd.Categorical(
    df['pclass'],
    categories=['1st', '2nd', '3rd'],
    ordered=True
)
print("\n--- Ket qua ---")
print(df.head())
print(df.info())
```
