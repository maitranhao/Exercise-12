```
import pandas as pd
import statsmodels.formula.api as smf

df_titanic = pd.read_csv('titanic.csv')

df_titanic = df_titanic.dropna(subset=['age', 'embarked'])

cong_thuc_titanic = 'survived ~ C(pclass) + sex + age + sibsp + C(embarked)'
model_logit = smf.logit(cong_thuc_titanic, data=df_titanic).fit(maxiter=100)

print("--- KẾT QUẢ TITANIC ---")
print(model_logit.summary())

df_wine = pd.read_csv('wines.csv')

cong_thuc_wine = """
quality ~ type + fixed_acidity + volatile_acidity + citric_acid +
residual_sugar + chlorides + free_sulfur_dioxide + total_sulfur_dioxide +
density + pH + sulphates + alcohol
"""

print("\n--- KẾT QUẢ WINE 2.1 ---")
model_ols = smf.ols(cong_thuc_wine, data=df_wine).fit()
print(model_ols.summary())

print("\n--- KẾT QUẢ WINE 2.3---")
try:
    model_mnlogit = smf.mnlogit(cong_thuc_wine, data=df_wine).fit(method='bfgs', maxiter=1000)
    print(model_mnlogit.summary())
except Exception as e:
    print("lỗi:", e)
```
